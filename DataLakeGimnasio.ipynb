```python
# 1. Importar librerías y conectar con el "Data Lake" (Google Drive)
from google.colab import drive
import pandas as pd
import random

# Montar Drive (esto conecta la máquina virtual con vuestra nube)
drive.mount('/content/drive')

# 2. Generar datos sintéticos del gimnasio (50000 clientes)
n_filas = 50000
data = {
    'Age': [random.randint(18, 65) for _ in range(n_filas)],
    'Contract_period': [random.choice([1, 6, 12]) for _ in range(n_filas)],
    'Avg_class_frequency_total': [round(random.uniform(0.5, 5.0), 2) for _ in range(n_filas)],
    'Churn': [random.choice([0, 1]) for _ in range(n_filas)]
}

df = pd.DataFrame(data)

# 3. Guardar el archivo directamente en la raíz de vuestro Google Drive
ruta_archivo = '/content/drive/My Drive/datos_gimnasio_cloud.csv'
df.to_csv(ruta_archivo, index=False)

print(f"✅ ¡Éxito! Archivo guardado en tu nube como: datos_gimnasio_cloud.csv")
```

```python
import time

inicio = time.time()

# Calculamos el promedio de la columna Age de golpe
promedio_python = df['Age'].mean()

fin = time.time()
tiempo_ms = (fin - inicio) * 1000

print(f"🏁 Python - Promedio de edad: {promedio_python:.2f}")
print(f"⏱️ Python - Tiempo invertido: {tiempo_ms:.2f} milisegundos.")
```
